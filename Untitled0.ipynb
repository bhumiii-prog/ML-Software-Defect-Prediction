```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler, LabelEncoder
import warnings
warnings.filterwarnings('ignore')

sns.set_style("whitegrid")
plt.rcParams['figure.dpi'] = 100
print("Libraries imported successfully!")
```

```python
url = "https://raw.githubusercontent.com/klainfo/DefectData/master/predictors/kc1.csv"

try:
    df = pd.read_csv(url)
    print("Loaded from URL")
except:
    print("Failed to load from URL")
    from google.colab import files
    uploaded = files.upload()
    filename = list(uploaded.keys())[0]
    df = pd.read_csv(filename)
    print(" Loaded from upload")

raw = df.copy()  # Keep original for before/after comparison
print(f"\nDataset Shape: {df.shape}")
print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe(include='all'))
```

```python
# Check missing values

missing_values = df.isnull().sum()

print("Missing Values in Dataset:")
print(missing_values)

print("\nTotal Missing Values:")
print(missing_values.sum())
```

```python
df.replace(
    ['########', '?', 'null', 'NULL', 'NA', ''],
    np.nan,
    inplace=True
)

for col in df.columns:
    if col != 'defects':
        df[col] = pd.to_numeric(df[col], errors='coerce')

print("Invalid values handled successfully!")

print("\nMissing Values After Conversion:")
print(df.isnull().sum())
```

```python
plt.figure(figsize=(14, 6))

sns.heatmap(
    df.isnull(),
    cbar=False,
    cmap='viridis'
)

plt.title(
    'Missing Values Before Preprocessing',
    fontsize=16
)

plt.xlabel('Features')
plt.ylabel('Records')

plt.tight_layout()
plt.show()
```

```python
df_before = df.copy()

numeric_columns = df.select_dtypes(
    include=np.number
).columns

for col in numeric_columns:

    df[col] = df[col].fillna(
        df[col].median()
    )

print("Missing values handled successfully!")

print("\nRemaining Missing Values:")
print(df.isnull().sum())
```

```python
# Compare missing values

fig, axes = plt.subplots(
    1, 2,
    figsize=(16, 6)
)

sns.heatmap(
    df_before.isnull(),
    cmap='viridis',
    cbar=False,
    ax=axes[0]
)

axes[0].set_title(
    'Before Missing Value Treatment'
)

sns.heatmap(
    df.isnull(),
    cmap='viridis',
    cbar=False,
    ax=axes[1]
)

axes[1].set_title(
    'After Missing Value Treatment'
)

plt.tight_layout()
plt.show()
```

```python
duplicate_count = df.duplicated().sum()

print("Total Duplicate Records:")
print(duplicate_count)
```

```python
rows_before = len(df)

df = df.drop_duplicates().copy()

rows_after = len(df)

print("Rows Before:", rows_before)
print("Rows After:", rows_after)

print("Duplicates Removed:",
      rows_before - rows_after)
```

```python
labels = ['Before Cleaning', 'After Cleaning']

values = [rows_before, rows_after]

plt.figure(figsize=(8, 5))

bars = plt.bar(
    labels,
    values,
    color=['tomato', 'mediumseagreen']
)

plt.title(
    'Dataset Size Before and After Duplicate Removal'
)

plt.ylabel('Number of Records')

for bar in bars:

    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        int(bar.get_height()),
        ha='center',
        va='bottom'
    )

plt.show()
```

```python
df['defects'] = (
    df['defects']
    .astype(str)
    .str.strip()
    .str.upper()
)

print("Target values before encoding:")

print(df['defects'].value_counts())

# Encode target

df['defects'] = df['defects'].map({
    'TRUE': 1,
    'FALSE': 0
})

print("\nTarget values after encoding:")

print(df['defects'].value_counts(dropna=False))
```

```python
print("Unrecognized target values:")
print(df['defects'].isnull().sum())

df = df.dropna(subset=['defects']).copy()

df['defects'] = df['defects'].astype(int)
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x='defects',
    hue='defects',
    palette=['skyblue', 'salmon'],
    legend=False
)

plt.title(
    'Distribution of Software Defects',
    fontsize=15
)

plt.xlabel('Defect Status')

plt.ylabel('Number of Software Modules')

plt.xticks(
    [0, 1],
    ['No Defect (0)', 'Defect (1)']
)

plt.show()
```

```python
features = ['loc', 'v(g)', 'v', 'e']

fig, axes = plt.subplots(
    2, 2,
    figsize=(14, 10)
)

axes = axes.flatten()

for i, col in enumerate(features):

    sns.boxplot(
        y=df[col],
        ax=axes[i],
        color='orange'
    )

    axes[i].set_title(
        f'Outlier Detection: {col}'
    )

plt.tight_layout()
plt.show()
```

```python
df_before_outliers = df.copy()

print("Outlier treatment started...")

numeric_features = df.select_dtypes(
    include=np.number
).columns.drop('defects')

for col in numeric_features:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df[col] = df[col].clip(
        lower=lower_bound,
        upper=upper_bound
    )

print("Outlier treatment completed!")
```

```python
fig, axes = plt.subplots(
    2, 4,
    figsize=(20, 10)
)

for i, col in enumerate(features):

    sns.boxplot(
        y=df_before_outliers[col],
        ax=axes[0, i],
        color='salmon'
    )

    axes[0, i].set_title(
        f'Before: {col}'
    )

    sns.boxplot(
        y=df[col],
        ax=axes[1, i],
        color='lightgreen'
    )

    axes[1, i].set_title(
        f'After: {col}'
    )

plt.tight_layout()
plt.show()
```

```python
X = df.drop('defects', axis=1)
y = df['defects']

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

df_scaled = pd.DataFrame(
    X_scaled,
    columns=X.columns,
    index=X.index
)

df_scaled['defects'] = y

print("Feature Scaling Completed!")

display(df_scaled.head())
```

```python
selected_features = [
    'loc',
    'v(g)',
    'v',
    'e'
]

fig, axes = plt.subplots(
    1, 2,
    figsize=(16, 6)
)

sns.boxplot(
    data=df[selected_features],
    ax=axes[0]
)

axes[0].set_title(
    'Before Feature Scaling'
)

axes[0].tick_params(
    axis='x',
    rotation=45
)

sns.boxplot(
    data=df_scaled[selected_features],
    ax=axes[1]
)

axes[1].set_title(
    'After Feature Scaling'
)

axes[1].tick_params(
    axis='x',
    rotation=45
)

plt.tight_layout()
plt.show()
```
